```python
# --- DETECTION: find rows whose STORED tenure contradicts the onboarding date ---

# Turn each row's stored tenure (in months) into a day-duration.
# ×30.44 = average days per month (365.25 ÷ 12); to_timedelta makes it a real span pandas can add to a date.
as_off = pd.to_timedelta(cust["wallet_tenure_months"] * 30.44, unit="D")

# Add that span to onboarding: "where does onboarding + stored tenure land?"
# For good rows this lands ~1 July 2025; for bad rows it overshoots into 2029–2031.
the_date = cust["onboarding_date"] + as_off

# --- CORRECT VALUE: compute the TRUE tenure from the confirmed as-of date ---

# The dataset's real "now" — reverse-engineered and validated earlier.
anchor = pd.Timestamp("2025-07-1")

# Days between onboarding and the as-of date. .dt.days pulls the whole-day count out of the duration.
gap_days = (anchor - cust["onboarding_date"]).dt.days

# Convert those days back to whole months → the correct tenure each row SHOULD have.
number = (gap_days / 30.44).round()

# --- THE CEILING: as-of date + ~3 weeks grace, absorbing 30.44 approximation wobble ---
searcher = pd.Timestamp("2025-07-22")

# Count rows whose implied end-date overshoots the ceiling = contradictory stored tenures (358).
print((the_date > searcher).sum())

# --- FIX: overwrite ONLY the contradictory rows with the correct computed tenure ---
# .loc[mask, column] targets just the flagged rows; number[mask] supplies each one's correct value.
# The ~14,642 good rows are untouched.
cust.loc[the_date > searcher, "wallet_tenure_months"] = number[the_date > searcher]
```

```python
# --- VALIDATION GATE: every assert encodes something the cleaning made true ---

# 1. Identity: one row per customer
assert cust["customer_id"].is_unique, "customer_id not unique"

# 2. No missing values where we fixed them (churned excluded — it's the target)
assert cust["age"].notna().all(), "age still has NaN"
assert cust["wallet_tenure_months"].notna().all(), "tenure still has NaN"
assert cust["declared_income_band"].notna().all(), "income-band still has NaN"

# 3. Ranges: imputed/cleaned values sit in sane bounds
assert cust["age"].between(18, 100).all(), "age out of range"
assert cust["wallet_tenure_months"].between(0, 120).all(), "tenure out of range"

# 4. Dates: nobody onboarded after the dataset's as-of date
assert cust["onboarding_date"].max() <= pd.Timestamp("2025-07-22"), "future onboarding"

# 5. Categoricals: only allowed values survived
assert set(cust["region"].unique()) <= {"Punjab","Sindh","KP","Balochistan","Islamabad","AJK-GB"}, "invalid region"
assert set(cust["segment_true"].unique()) <= {"payroll","merchant","borrower","saver"}, "invalid segment"

# 6. Booleans are truly 0/1
for col in ["smartphone_user","has_savings","has_insurance","is_whale"]:
    assert set(cust[col].unique()) <= {0,1}, f"{col} not binary"

print("All validation checks passed ✅")
```
